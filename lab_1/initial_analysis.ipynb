```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('ForeignGifts_edu.csv',low_memory=False)
df.head()
```

```python
#B. Describe the differences between different classes of gift types.
df['Gift Type'].unique()
df['Gift Type'].value_counts(normalize=True)
df['Gift Type'].hist( bins = 35 )
#The histogram shows that the most common gift type is "Contract", the second is "Monetary Gift", and the least common is "Real Estate", being more skewed in favor of two gift types rather than all.
```

```python
#Question 1: Which country gives the most money, in total?
df[['Country of Giftor','Foreign Gift Amount']].groupby('Country of Giftor').sum().sort_values(by='Foreign Gift Amount',ascending=False)
#Qatar gives the most money in total with $2706240869.
```

```python
#Question 2: Which country initiates the most gifts, by count?
df[['Country of Giftor','Foreign Gift Amount']].groupby('Country of Giftor').count().sort_values(by='Foreign Gift Amount',ascending=False)
#England initiates the most gifts by count at 3655.
```

```python
#Question 3: Which country gives the largest gifts, on average?
df[['Country of Giftor','Foreign Gift Amount']].groupby('Country of Giftor').mean().sort_values(by='Foreign Gift Amount',ascending=False)
#Bermuda gives the largest gifts on average at $7688837
```

```python
#Question 4: Which institution receive the most money, in total?
df[['Institution Name','Foreign Gift Amount']].groupby('Institution Name').sum().sort_values(by='Foreign Gift Amount',ascending=False).head(20)
# Carnegie Mellon University receives the most money in total at $1477922504
```

```python
#Question 5: Which institution receive the greatest number of gifts, by count?
df[['Institution Name','Foreign Gift Amount']].groupby('Institution Name').count().sort_values(by='Foreign Gift Amount',ascending=False).head(20)
# University of California receives the most gifts by count at 3916
```

```python
#Build a plot of the top 20 institutions by the amount or number of gifts they have received using either seaborn or plotly
sns.barplot(data=df[['Institution Name','Foreign Gift Amount']].groupby('Institution Name').count().sort_values(by='Foreign Gift Amount',ascending=False).head(20), x='Institution Name', y='Foreign Gift Amount')
plt.xticks(rotation=45, ha='right')
```

```python
#Question 6: What is the average amount each university receives? Median amount? If they are different why?
df[['Institution Name','Foreign Gift Amount']].groupby('Institution Name').mean().sort_values(by='Foreign Gift Amount',ascending=False)
```

```python
#Question 6: What is the average amount each university receives? Median amount? If they are different why?
df[['Institution Name','Foreign Gift Amount']].groupby('Institution Name').median().sort_values(by='Foreign Gift Amount',ascending=False)
#There are differences between the average and median amounts because of outlier that influence the average more than the median.
```

```python
university_averages = df[['Institution Name','Foreign Gift Amount']].groupby('Institution Name').mean()

plt.hist(university_averages['Foreign Gift Amount'], bins=35)

plt.axvline(university_averages['Foreign Gift Amount'].mean(),
            linestyle='--', color='red', label='Mean')

plt.axvline(university_averages['Foreign Gift Amount'].median(),
            linestyle='--', color='black', label='Median')

plt.xlabel('Average Gift Amount ($)')
plt.ylabel('Number of Universities')
plt.title('Average Foreign Gift Amount by University')
plt.legend()
plt.show()
```

```python
#Question 7: What is the largest flow of a single country to a single institution?
giftor = 'Country of Giftor'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
df[['Country of Giftor', 'Institution Name', 'Foreign Gift Amount']].groupby(['Country of Giftor', 'Institution Name']).sum().sort_values(by='Foreign Gift Amount', ascending=False).head(1)
#The largest flow is Qatar to Cornell University.
```

```python

import plotly.graph_objects as go

giftor = 'Country of Giftor'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
N = 25

flows = (
    df.groupby([giftor, 
                recipi])
      [flow]
      .sum()
      .nlargest(N)
      .reset_index()
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[flow]
        )
    )
)

fig.show()
#Question 8: Which are the top giftors to US academic institutions?
#Qatar, England, and China are the top 3 giftors to US academic institutions.
```
