```python
peers = [
    'College of William & Mary',
    'George Mason University',
    'Virginia Commonwealth University',
    'Virginia Polytechnic Institute & State University'
]
compare = df[df['Institution Name'].isin(peers + ['University of Virginia'])]
comparetotal = compare.groupby('Institution Name')['Foreign Gift Amount'].sum().sort_values(ascending=False)
comparetotal
```

```python
comparemean = compare.groupby('Institution Name')['Foreign Gift Amount'].mean().sort_values(ascending=False)
comparemean
```

```python
market_total = df['Foreign Gift Amount'].sum()
market_count = df['Foreign Gift Amount'].count()
market_average = df['Foreign Gift Amount'].mean()
print("Market total:", market_total) 
print("Market count:", market_count)
print("Market average:", market_average)
```

```python
Total_UVA = UVA['Foreign Gift Amount'].sum()
UVA_share = Total_UVA / market_total
print("UVA percentage share:", UVA_share * 100, "%")
```

```python
#I believe UVA should try to target countries that donate a lot to USA universities in general but don't target UVA, in order to capitilize on the interest in donating to higher education by those countries.
UVA_foreigndonation = UVA[['Country of Giftor','Foreign Gift Amount']].groupby('Country of Giftor').sum()
Overall_foreigndonation = df[['Country of Giftor','Foreign Gift Amount']].groupby('Country of Giftor').sum().sort_values(by='Foreign Gift Amount', ascending=False).head(10)

comparison = pd.DataFrame({
    'Market Total': Overall_foreigndonation['Foreign Gift Amount'],
    'UVA Total': UVA_foreigndonation['Foreign Gift Amount']
}, index=Overall_foreigndonation.index).fillna(0)

comparison['UVA Market Share (%)'] = comparison['UVA Total'] / comparison['Market Total'] * 100
comparison
```
